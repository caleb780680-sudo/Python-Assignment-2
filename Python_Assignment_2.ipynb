```python
# 1. List Creation
age_list = [24, 25, 27, 28, 29]
name_list = ["Zara", "Ravi", "Priya", "Kumar", "Sita"]

# 2. List Operations / Modifications
# a. Append the string "Yazhini" to name_list
name_list.append("Yazhini")

# b. Insert the element 26 at index 2 in age_list
age_list.insert(2, 26)

# c. Remove the string "Yazhini" from name_list
name_list.remove("Yazhini")

# d. Pop the last element from age_list
age_list.pop()

# e. Extend the age_list with additional ages [31, 30, 32]
age_list.extend([31, 30, 32])

# f. Sort age_list in descending order
age_list.sort(reverse=True)

# g. Find the max age, min age, and sum of all ages from age_list
max_age = max(age_list)
min_age = min(age_list)
sum_age = sum(age_list)

print("Updated age_list:", age_list)
print("Updated name_list:", name_list)
print("Max Age:", max_age, "| Min Age:", min_age, "| Sum of Ages:", sum_age)

# 3. Accessing List Elements
# a. Print the first element of name_list
print("a. First element:", name_list[0])

# b. Print the last element of name_list
print("b. Last element:", name_list[-1])

# c. Print the elements from index 2 to index 4 in name_list
print("c. Elements from index 2 to 4:", name_list[2:5])

# d. Print the elements of name_list in reverse order
print("d. Reversed name_list:", name_list[::-1])
```

```python
# a. Create a dictionary named student_marks
student_marks = {
    "Zara": 85,
    "Ravi": 78,
    "Priya": 92,
    "Kumar": 67,
    "Sita": 88
}

# b. Access and print the mark of a specific student of your choice
print("b. Mark of Priya:", student_marks["Priya"])

# c. Add a new student, "Janani", with a mark of 80
student_marks["Janani"] = 80

# d. Update the mark of any one older student to 82
student_marks["Ravi"] = 82

# e. Use keys(), values(), and items() methods
print("e. Keys:", list(student_marks.keys()))
print("e. Values:", list(student_marks.values()))
print("e. Items (Key-Value pairs):", list(student_marks.items()))
```

```python
# a. Create a set called my_set
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print("a. my_set output:", my_set)
# Explanation: Sets automatically remove duplicate elements and store unique values only.

# b. Attempt to access my_set[4]
try:
    print(my_set[4])
except TypeError as e:
    print("b. Error Explanation: Sets are unordered collections, so they do not support indexing.")

# c & d. Sets union and intersection
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

union_set = set1.union(set2)
intersection_set = set1.intersection(set2)

print("d. Union of set1 and set2:", union_set)
print("d. Intersection of set1 and set2:", intersection_set)
```

```python
# Performance Category Program

# 1. Prompt the user for input
score = float(input("Enter your score (0 to 10): "))

# 2 & 3 & 4. Find category, print output, and provide feedback
if score > 7:
    category = "Above Average"
    feedback = "Excellent performance! Keep up the great work."
elif 4 <= score <= 7:
    category = "Average"
    feedback = "Good effort! Keep practicing, there's room for improvement."
else:
    category = "Below Average"
    feedback = "Need to improve your performance; consistent practice will lead to better results."

print(f"Performance Category: {category}")
print(f"Message: {feedback}")
```
